## Waarom deze modellen?

De mediaan is de eerlijke naïeve referentie. Ridge test een eenvoudige lineaire relatie.
Histogram Gradient Boosting vangt niet-lineaire interacties efficiënt. Extra Trees kan
complexere stations- en tijdspatronen leren, maar overfit makkelijker. CatBoost verwerkt
stations als echte categorieën. Zowel een log1p-target als directe duurtraining worden
vergeleken. De afstand en uitsluitend uit eerdere ritten opgebouwde routehistorie geven
routecontext zonder informatie uit de toekomst te lekken.

## AutoML (FLAML)

FLAML vervangt hier PyCaret als toegestane AutoML-vergelijking. Voor een eerlijke vergelijking krijgt FLAML exact dezelfde 640k selectie-trainingsritten, 160k chronologische validatieritten, features en MAE als de handmatige modellen. De aparte finale hold-out blijft buiten AutoML. FLAML vergelijkt vijf op Windows/Python 3.13 geteste modeltypen (`histgb`, `extra_tree`, `rf`, `catboost` en `xgboost`). LightGBM is uitgesloten omdat de native wheel in deze runtime reproduceerbaar crasht; het tijdsbudget maakt de overige zoekactie reproduceerbaar qua kosten.

```python
import json
import pandas as pd

RUN_AUTOML = False  # Zet alleen op True om de opgeslagen 15-minutenrun te herhalen.
if RUN_AUTOML:
    from src.models.automl import run_automl
    run_automl(
        PROCESSED_DIR / "trips.parquet", seconds=900, max_rows=1_000_000,
        trial_seconds=180,
    )

automl_result = json.loads((REPORT_DIR / "automl_result.json").read_text())
automl_metrics = pd.read_csv(REPORT_DIR / "automl_metrics.csv")
display(automl_metrics.style.format({"mae_minutes": "{:.2f}"}))
display(automl_result)
```

## Selectie

De vergelijking gebruikt voor beide routes exact 639.865 trainingsritten en dezelfde 159.967 validatieritten (5 april 2024–30 mei 2025). Alle vijf FLAML-learners zijn geprobeerd. FLAML koos XGBoost met MAE **4,48 minuten**; de handmatige `hist_gradient_boosting_absolute` behaalt **3,62 minuten**. Het bestaande Histogram Gradient Boosting-model blijft dus de winnaar. Omdat FLAML niet won, is de gereserveerde finale hold-out terecht niet opnieuw gebruikt voor een tweede modelkeuze.
